```python
# Identify problematic TotalCharges records

total_numeric = pd.to_numeric(
    data["TotalCharges"],
    errors="coerce"
)

invalid_mask = total_numeric.isna()

problem_records = data.loc[
    invalid_mask,
    ["customerID", "tenure", "MonthlyCharges",
     "TotalCharges", "Churn"]
]

display(problem_records)
```

```python
# Summarize tenure of problematic records

print(problem_records["tenure"].value_counts())

print(
    "\nAll problematic records have zero tenure:",
    (problem_records["tenure"] == 0).all()
)
```

```python
data_clean = data.copy()
print("Orginal Dataset:",data.shape)
print("Cleaned Working Copy", data_clean.shape)
```

```python
data_clean["TotalCharges"] = (data_clean["TotalCharges"].str.strip())

data_clean["TotalCharges"] = pd.to_numeric(data_clean["TotalCharges"],errors="coerce")

print(data_clean["TotalCharges"].dtype)
```

```python
zero_tenure_missing = ((data_clean["tenure"] == 0) & (data_clean["TotalCharges"].isna()))
print(zero_tenure_missing)
```

```python
data_clean.loc[zero_tenure_missing,"TotalCharges"] = 0.0
```

```python
# Verify TotalCharges
print("Data type:", data_clean["TotalCharges"].dtype)

print(
    "Missing TotalCharges:",
    data_clean["TotalCharges"].isna().sum()
)

print(
    "Zero TotalCharges:",
    (data_clean["TotalCharges"] == 0).sum()
)

print("Dataset shape:", data_clean.shape)
```

```python
# Check missing values in the cleaned dataset

missing_after_cleaning = data_clean.isnull().sum()

print("Missing values after cleaning:")
print(missing_after_cleaning[missing_after_cleaning > 0])

print(
    "\nTotal missing values:",
    data_clean.isnull().sum().sum()
)
```

```python
# Verify duplicate records

print(
    "Duplicate records:",
    data_clean.duplicated().sum()
)
```

```python
# Display all column data types

print(data_clean.dtypes)
```

```python
# Inspect important categorical columns

categorical_columns = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "InternetService",
    "Contract",
    "PaymentMethod",
    "Churn"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(data_clean[column].value_counts(dropna=False))
```

```python
# Final dataset validation

print("=" * 45)
print("CLEANED DATASET VALIDATION")
print("=" * 45)

print("Rows:", data_clean.shape[0])
print("Columns:", data_clean.shape[1])

print("Total missing values:", data_clean.isnull().sum().sum())

print("Duplicate records:", data_clean.duplicated().sum())

print("TotalCharges data type:", data_clean["TotalCharges"].dtype)

print("Zero TotalCharges:", (data_clean["TotalCharges"] == 0).sum())

print("Original dataset unchanged:", data.shape == data_clean.shape)
```

```python
# Define the output file path
output_path = "../data/processed/telco_churn_cleaned.csv"

# Save the cleaned dataset
data_clean.to_csv(
    output_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print("File location:", output_path)
```

```python
from pathlib import Path

saved_file = Path(output_path)

print("File exists:", saved_file.exists())

if saved_file.exists():
    print("File size:", round(saved_file.stat().st_size / 1024, 2), "KB")
```
